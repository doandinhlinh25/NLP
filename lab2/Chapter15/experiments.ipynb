## 2. Tính sentence probability và sentence log probability

```python
training_corpus = [
    tokenize('the cat eats fish'),
    tokenize('the cat likes fish'),
    tokenize('the dog eats meat'),
    tokenize('the cat eats fish the cat eats fish'),
]
model = NGramLanguageModel(order=2).fit(training_corpus)
sentence = tokenize('the cat eats fish')
sentence_probability = model.sentence_probability(sentence)
sentence_log_probability = model.sentence_log_probability(sentence)
display({
    'sentence': 'the cat eats fish',
    'probability': sentence_probability,
    'log_probability': sentence_log_probability,
    'exp(log_probability)': exp(sentence_log_probability),
})
```

## 3. Minh họa numerical underflow với câu dài

```python
long_sentence = sentence * 2_000
display({
    'token_count': len(long_sentence),
    'direct_probability': model.sentence_probability(long_sentence),
    'log_probability': model.sentence_log_probability(long_sentence),
})
```
